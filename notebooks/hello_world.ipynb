```python
import sys
print("Hello World depuis", sys.executable)
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("hello-world").getOrCreate()
print("Spark", spark.version)
```

```python
df = spark.createDataFrame([(1, "Hello"), (2, "World")], ["id", "mot"])
df.show()
```

## Écriture / lecture Parquet dans HDFS

```python
chemin = "/retail-lake/tmp/hello_world.parquet"
df.write.mode("overwrite").parquet(chemin)
spark.read.parquet(chemin).show()
```
